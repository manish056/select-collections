```python
import warnings
warnings.simplefilter(action='ignore', category=UserWarning)

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, OneHotEncoder, PowerTransformer
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, VotingRegressor
from sklearn.linear_model import LinearRegression, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.metrics import mean_squared_error
from sklearn.feature_selection import SelectFromModel
import matplotlib.pyplot as plt
import xgboost as xgb
```

```python
# Load the original data
file_path = r"C:\Users\manis\Downloads\UV6413-XLS-ENG.xlsx"
data = pd.read_excel(file_path, sheet_name='Training Data')

print("Original data shape:", data.shape)
print(data.describe())
```

```python
# Feature engineering
data['payment_ratio'] = data['totalpay'] / (data['cobal'] + 1)
data['bureau_per_call'] = data['bureauscr'] / (data['numcalls'] + 1)
data['balance_bureau_ratio'] = data['cobal'] / (data['bureauscr'] + 1)
data['log_cobal'] = np.log1p(data['cobal'])
data['log_numcalls'] = np.log1p(data['numcalls'])
data['sqrt_cobal'] = np.sqrt(data['cobal'])
data['sqrt_numcalls'] = np.sqrt(data['numcalls'])
data['cobal_squared'] = data['cobal'] ** 2
data['bureauscr_squared'] = data['bureauscr'] ** 2
data['bureau_bin'] = pd.qcut(data['bureauscr'], q=5, labels=False, duplicates='drop')
data['cobal_bureau'] = data['cobal'] * data['bureauscr']
data['calls_bureau'] = data['numcalls'] * data['bureauscr']

print("\nData shape after feature engineering:", data.shape)
```

```python
# Define features and target
X = data.drop(columns=['totalpay'])
y = data['totalpay']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

categorical_cols = ['state', 'rollout']
numeric_cols = [col for col in X.columns if col not in categorical_cols]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', PowerTransformer(), numeric_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ])

# Feature selection function
def select_important_features(X_train_processed, y_train, threshold=0.01):
    selector = SelectFromModel(
        RandomForestRegressor(n_estimators=100, random_state=42), 
        threshold=threshold
    )
    selector.fit(X_train_processed, y_train)
    return selector

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

selector = select_important_features(X_train_processed, y_train)
X_train_selected = selector.transform(X_train_processed)
X_test_selected = selector.transform(X_test_processed)

print(f"Selected {X_train_selected.shape[1]} features out of {X_train_processed.shape[1]}")
```

```python
rf_model = RandomForestRegressor(n_estimators=200, max_depth=15, min_samples_split=5,
                                  min_samples_leaf=2, max_features='sqrt', bootstrap=True, random_state=42)
rf_model.fit(X_train_selected, y_train)
rf_pred = rf_model.predict(X_test_selected)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))

xgb_model = xgb.XGBRegressor(n_estimators=300, learning_rate=0.05, max_depth=8,
                             subsample=0.8, colsample_bytree=0.8, gamma=0.1, random_state=42)
xgb_model.fit(X_train_selected, y_train)
xgb_pred = xgb_model.predict(X_test_selected)
xgb_rmse = np.sqrt(mean_squared_error(y_test, xgb_pred))

gb_model = GradientBoostingRegressor(n_estimators=250, learning_rate=0.05, max_depth=6,
                                     min_samples_split=4, min_samples_leaf=2, random_state=42)
gb_model.fit(X_train_selected, y_train)
gb_pred = gb_model.predict(X_test_selected)
gb_rmse = np.sqrt(mean_squared_error(y_test, gb_pred))

dt_model = DecisionTreeRegressor(max_depth=10, random_state=42)
dt_model.fit(X_train_selected, y_train)
dt_pred = dt_model.predict(X_test_selected)
dt_rmse = np.sqrt(mean_squared_error(y_test, dt_pred))

lr_model = LinearRegression()
lr_model.fit(X_train_selected, y_train)
lr_pred = lr_model.predict(X_test_selected)
lr_rmse = np.sqrt(mean_squared_error(y_test, lr_pred))

lasso_model = Lasso(alpha=0.1, max_iter=10000, random_state=42)
lasso_model.fit(X_train_selected, y_train)
lasso_pred = lasso_model.predict(X_test_selected)
lasso_rmse = np.sqrt(mean_squared_error(y_test, lasso_pred))

ensemble_model = VotingRegressor([
    ('rf', rf_model),
    ('xgb', xgb_model),
    ('gb', gb_model),
    ('dt', dt_model),
    ('lr', lr_model),
    ('lasso', lasso_model)
])
ensemble_model.fit(X_train_selected, y_train)
ensemble_pred = ensemble_model.predict(X_test_selected)
ensemble_rmse = np.sqrt(mean_squared_error(y_test, ensemble_pred))

weighted_pred = (
    0.20 * rf_pred + 
    0.25 * xgb_pred + 
    0.20 * gb_pred + 
    0.10 * dt_pred + 
    0.15 * lr_pred + 
    0.10 * lasso_pred
)
weighted_rmse = np.sqrt(mean_squared_error(y_test, weighted_pred))

# Print RMSEs
print(f"RandomForest RMSE: {rf_rmse}")
print(f"XGBoost RMSE: {xgb_rmse}")
print(f"GradientBoosting RMSE: {gb_rmse}")
print(f"DecisionTree RMSE: {dt_rmse}")
print(f"LinearRegression RMSE: {lr_rmse}")
print(f"Lasso RMSE: {lasso_rmse}")
print(f"VotingEnsemble RMSE: {ensemble_rmse}")
print(f"Weighted Ensemble RMSE: {weighted_rmse}")
```

```python
# Create prediction DataFrame
results_df = pd.DataFrame({
    'Actual': y_test,
    'RF_Predicted': rf_pred,
    'XGB_Predicted': xgb_pred,
    'GB_Predicted': gb_pred,
    'DT_Predicted': dt_pred,
    'LR_Predicted': lr_pred,
    'Lasso_Predicted': lasso_pred,
    'VotingEnsemble': ensemble_pred,
    'Predicted_TotalPay': weighted_pred
})
results_df = results_df.reset_index(drop=True)

# RMSE Summary Table
rmse_summary = pd.DataFrame({
    'Model': ['RandomForest', 'XGBoost', 'GradientBoosting', 'DecisionTree', 'LinearRegression', 'Lasso', 'VotingEnsemble', 'WeightedEnsemble'],
    'RMSE': [rf_rmse, xgb_rmse, gb_rmse, dt_rmse, lr_rmse, lasso_rmse, ensemble_rmse, weighted_rmse]
})
```

```python
# Merge original test set with predictions
X_test_with_index = X_test.copy()
X_test_with_index['totalpay'] = y_test.values
merged_df = pd.concat([X_test_with_index.reset_index(drop=True), results_df.drop(columns='Actual')], axis=1)

# Save to Excel
final_output_file = 'final_merged_predictions_output.xlsx'
with pd.ExcelWriter(final_output_file) as writer:
    merged_df.to_excel(writer, index=False, sheet_name='Merged Predictions')
    rmse_summary.to_excel(writer, index=False, sheet_name='RMSE Summary')

print(f"\n Final Excel file saved as: {final_output_file}")
```
